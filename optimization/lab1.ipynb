```python
from scalar_optimization import *
```

```python
a = 12
b = 20
f = lambda x: ((x - 15)**2) + 5
```

```python
result = golden_ratio(f, a, b, epsilon=0.01, verbose=True)
print(f"\nРезультат: x* = {result}")
print(f"f(x*) = {round(f(result), 2)}")
```

```python
result = dichotomy(f, a, b, epsilon=0.01, verbose=True)
print(f"\nРезультат: x* = {result}")
print(f"f(x*) = {round(f(result), 2)}")
```

```python
result = fibonacci(f, a, b, epsilon=0.01, verbose=True)
print(f"\nРезультат: x* = {result}")
print(f"f(x*) = {round(f(result), 2)}")
```
